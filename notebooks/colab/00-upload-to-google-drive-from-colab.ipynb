## Download the Digikala dataset onto Drive

Source: [RadeAI/Digikala_comments_products](https://huggingface.co/datasets/RadeAI/Digikala_comments_products/tree/main)

The two tables on that page are saved under `MyDrive/digikala-products-recommender-systems/data/raw`:

- `digikala-comments.csv` — 1,278,526,959 bytes
- `digikala-products.csv` — 258,733,765 bytes

`digikala.rar` on the same page is the packed copy of this release, so it is not downloaded again. A file already on Drive with the expected size is left in place.

```python
!pip install -q -U huggingface_hub
```

```python
import os
import shutil

from huggingface_hub import hf_hub_download

REPO = "RadeAI/Digikala_comments_products"
DEST = "/content/drive/MyDrive/digikala-products-recommender-systems/data/raw"
FILES = {
    "digikala-comments.csv": 1_278_526_959,
    "digikala-products.csv": 258_733_765,
}

os.makedirs(DEST, exist_ok=True)

for name, expected in FILES.items():
    target = os.path.join(DEST, name)
    if os.path.isfile(target) and os.path.getsize(target) == expected:
        print(f"already on Drive: {target} ({expected:,} bytes)")
        continue

    print(f"downloading {name} ...")
    cached = hf_hub_download(
        repo_id=REPO,
        repo_type="dataset",
        filename=name,
    )
    partial = target + ".partial"
    shutil.copyfile(cached, partial)
    os.replace(partial, target)

    got = os.path.getsize(target)
    if got != expected:
        raise RuntimeError(f"{name}: expected {expected:,} bytes, Drive has {got:,}")
    print(f"saved {target} ({got:,} bytes)")
```

```python
for name in FILES:
    path = os.path.join(DEST, name)
    with open(path, "r", encoding="utf-8", errors="replace") as handle:
        header = handle.readline().rstrip("\n")
    print(f"{name}: {os.path.getsize(path):,} bytes")
    print(header)
    print()
```
